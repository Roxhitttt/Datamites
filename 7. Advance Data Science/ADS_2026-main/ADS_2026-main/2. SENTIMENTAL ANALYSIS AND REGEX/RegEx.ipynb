```python
sen = 'I have list of email ids here. url@connect.com, apple@berry.com, web@browser.com. @Navin, @Pravin'
#Find the people who are tagged in this message 
print(re.findall(r' @([A-Za-z]+)',sen))

# Find the email ids listed in the message
print(re.findall(r' [a-zA-Z_0-9]+@[a-z]+\.com',sen))

# Check if the message end 'mails.'
print(re.findall('mails.$',sen))

# Check if the message starts with I or WE
print(re.findall('^(I|We)',sen)) # ^ indicates beginning of sentences and | indicates 'or'
```

```python
# If you want a Regex expression to validate an Indian phone number, use:
numbers = [
    '9876543210',
    '+919876543210',
    '+91 9876543210',
    '+91-9876543210'
]

pattern = r'^(\+91[-\s]?)?[6-9]\d{9}$'

for number in numbers:
    if re.fullmatch(pattern, number):
        print(number, "Valid")
    else:
        print(number, "Invalid")
```
