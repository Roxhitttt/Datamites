# Machine Learning I

**Machine Learning (ML)** provides statistical tools to explore data. It works only on **structured data**, not unstructured.

---

## 1. Introduction to ML

- **Machine learning** is a part/subset of **AI**  
- It’s the ability to **automatically learn & improve from experience**, without explicit programming  
- **Example**: Netflix  
  - You frequently watch horror movies  
  - It shows recommended movies based on your past viewing data  
- We **teach ML** to analyze past data and predict outputs **without explicit programming**

### Comparison with Traditional Programming

| Method              | Process                                                                 |
|---------------------|-------------------------------------------------------------------------|
| Traditional program | Contains explicit rule, e.g., a × b <br> Applies rule to inputs a, b to compute output |
| ML learner          | Learns the rule from data (a, b, output) <br> Given new input (3, 5), it predicts the output |

- The ML model *identifies the relationship* between input (x) and output (y)  
- After training, it can *predict the value* for different values of x

## 1. Supervised Learning Machine

- Input and output (labels) are provided  
- It uses *labeled data*  
- The model *predicts* output (e.g., classification, regression)  
- ML models are *trained* using labeled data  
- Once trained, it predicts outputs for new inputs

*Examples*: Regression & Classification

## 2. Unsupervised Learning

**Definition**:  
Unsupervised learning uses **unlabeled data**. The algorithm tries to find structure, patterns, or groupings in the input data without any provided outcomes.
**How it works**:
- No labels are provided
- The algorithm analyzes input features
- Groups similar items 

**Common Algorithms**:
- K-Means Clustering

# Types of Supervised Learning

Supervised Machine Learning algorithms can be broadly classified into **Regression** and **Classification** algorithms.

---

## 📈 Regression

- A **predictive modeling** technique.
- Regression analysis is a statistical method to model the relationship between:
  - A **dependent** (target) variable, and
  - One or more **independent** (predictor) variables.
- Helps understand how the value of the dependent variable changes when independent variables vary, holding others fixed.
- Predicts **continuous/real** values such as temperature, age, salary, price, etc.

**Common types of regression:**
1. Linear Regression  
2. Time Series Analysis  

---

## 🧩 Classification

- Used to **identify the category** of new observations based on training data.
- The model learns from existing data and classifies new observations into a number of classes or groups.
- Examples: Yes or No, 0 or 1, Spam or Not Spam, Cat or Dog, etc.
- These classes are referred to as **targets**, **labels**, or **categories**.

**Types of classification:**
1. Binary Classification  
2. Multiclass Classification  


# Linear Regression

## Overview
- Supervised learning technique  
- Used to predict **continuous values**  
- Finds the **linear relationship between variables**

### Variables
- **Input variable (X)** → Independent / Predictor variable  
- **Output variable (Y)** → Dependent / Response / Target variable  

### Equation
$$
y = m_1 x_1 + m_2 x_2 + m_3 x_3 + \cdots + c
$$

### Note
- Standardization (scaling $\mu = 0, \sigma = 1$) helps optimization  
- Works best when features are approximately normally distributed  

---

## Assumptions of Linear Regression

### 1. Linearity
Relationship between $X$ and $Y$ should be linear  


### 2. Independence (No Multicollinearity)
No strong correlation between independent variables  



---

## Objective of Linear Regression

### Best Fit Line
The best-fit line minimizes the **Residual Sum of Squares (RSS)**  

### Error
$$
\text{Error} = \text{Actual Value} - \text{Predicted Value}
$$

### Residual
$$
\text{Residual} = y - \hat{y}
$$

### Residual Sum of Squares (RSS)
$$
\text{RSS} = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

- Lower RSS → Better fit  

---

## Error Metrics

### 1. Mean Squared Error (MSE)
$$
\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

### 2. Mean Absolute Error (MAE)
$$
\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} \left|y_i - \hat{y}_i\right|
$$

### 3. Root Mean Squared Error (RMSE)
$$
\text{RMSE} = \sqrt{\text{MSE}}
$$

---

## Evaluating a Model

### R² Score (Coefficient of Determination)
- Measures how well data fits the regression line  

### Interpretation
- $R^2 = 1$ → Perfect model  
- $R^2 = 0$ → No predictive power  
- $0 < R^2 < 1$ → Partial explanation  
- $R^2 < 0$ → Worse than baseline  

**Example:**  
If $R^2 = 0.10$, then **10% of variance in $Y$ is explained by $X$**

### Formula
$$
R^2 = 1 - \frac{\text{RSS}}{\text{TSS}}
$$

Where:
$$
\text{TSS} = \sum_{i=1}^{n} (y_i - \bar{y})^2
$$

---

## Adjusted R² Score

### Definition
Modified version of $R^2$ that penalizes unnecessary features  

### Formula
$$
\text{Adjusted } R^2 = 1 - \frac{(1 - R^2)(n - 1)}{n - p - 1}
$$

Where:
- $n$ = number of data points  
- $p$ = number of predictors  

### Behavior
- Adding irrelevant feature → Adjusted $R^2$ decreases  
- Adding useful feature → Adjusted $R^2$ increases
