# Difference between list and tuple 

The main difference between lists and tuples in programming, particularly in Python, is their mutability.

* Lists are mutable, meaning you can change their contents (add, remove, or modify elements) after they are created. They are typically defined using square brackets ([]).

* Tuples are immutable, meaning you cannot change their contents once they are created. They are typically defined using parentheses (()).

# Sets 
A Set in Python is a collection of unique elements that is unordered and mutable.

Think of it like a bag of distinct items where the order you put them in doesn't matter, and duplicates are automatically thrown out.

## Key Features
* Unique Elements: Sets do not allow duplicate values. If you try to add a duplicate, it is ignored.

* Unordered: Items have no fixed position. You cannot access items using an index (e.g., myset[0] will raise an error).

* Mutable: You can add or remove items after the set is created.

* Syntax: Defined using curly braces {} or the set() function.

```python
# Creating a set
fruits = {"apple", "banana", "orange"}
# Adding elements
fruits.add("grape")
print(fruits)
# Removing elements 
fruits.remove("banana")
print(fruits)
```

```python
# Intersection of two lists, not to use typecast in sets

lst1 = [2,3,3,4,5]
lst2 = [3,4,5,6,7]
result = [] # Empty list
for num in lst1: # checking num in lst1 and num in lst2 but not in result then num will be added in result list
    if num in lst2 and num not in result:
        result.append(num)
print(result)
```

# Dictionaries

A Dictionary in Python is a collection that stores data in key-value pairs.

* Key-Value Structure: Every item consists of a pair: Key: Value.

* Fast Lookup: Designed to retrieve data very quickly using the key.

* Mutable: You can add, remove, or change items after creation.

* Unique Keys: Keys must be unique (no duplicates allowed) and immutable (e.g., strings, numbers). Values can be anything.

```python
intro = {"Name" : "Ram", "Age" : 3, "Gen" : "M"}
print(intro)
intro["City"] = "Indore" # Adding new key value pairs
print(intro)
intro["Name"] # to access the values

intro["Age"] = 4 # modifying value
print(intro)

del intro["Gen"]
print(intro)

intro.keys() # to access the keys 

intro.values() # to access the values 

intro.items() # to access the items
```

```python
intro.clear() # dictionary gets empty
print(intro)
```

```python
# Task - input msg, and then count of every character

msg = input("Emter your message : ")

checking_char = "" # This string will remember characters we have already printed
Total_count = 0

for i in msg:
    # Only count this character if we haven't seen it yet
    if i not in checking_char:
        count = msg.count(i)
        print(i, ":", count)
        # Add the character to our "checked" string so we skip it next time
        Total_count += 1 # CHecking_char = checking_char + 1
print(f"Total count of number : 
{Total_count}")
```

# Filter Function 

```python
# a list contains both even and odd numbers.
LST = [0,1,2,3,5,8,13]

# result contains odd numbers of the list
result = filter(lambda x: x % 2 == 0, LST)
result = list(result) # Type casting this result coz filter function contains it's own filter object output like this <filter object at 0x0000023615933CA0>
print(result)
```

```python
# Task : Create a fucntion which will caluclate average of a list

def avg(lst):
    total = 0
    count = 0
    for lst_avg in lst:
        total += lst_avg # total = total + lst_avg
        count += 1
    print(total/count)
avg([3, 4, 6, 7, 23])
```

```python
# Types of arguments

# 1. positional arguments - above 2 exammples of positional arguments
# 2. Keyword arguments : *args - can pass n number of values. args can be any word

# Example of Keyword arguments - sum of numbers

def sum_n(*values):
    total = 0
    for num in values:
        total += num
    print(total)
sum_n(23, 54, 31, 67, 34)
```

```python
# Create a function which takes a message as input and tells you the frequency of each vowel in it.

def frqvow():
    message = input("Enter a message : ")

    # we define the vowels we look for 
    vowel = "aeiou"

    print("frequency of each vowel : ") 

    # Loop through each vowel (a, e, i, o, u) one by one
    for v in vowel:
        # Count how many times this specific vowel appears in the message
        # .lower() ensures we count 'A' and 'a' together
        count = message.lower().count(v)

        # Only print if the vowel is actually present
        if count > 0:
            print(f"{v} : {count}")
# Call the function (no arguments needed)            
frqvow()
```

```python
# input a message and then count a frequncy each vowel present in the message

def cv(message):
    message = message.upper()
    result={} # Empty dictionary
    for char in message:
        if char in "AEIOU" and char not in result.keys(): # if char in "AEIOU": This effectively filters out all consonants and spaces
            result[char] = 1
        elif char in "AEIOU" and char in result.keys():
            result[char] += 1
    print(result)
cv("Hello, Harshit Master is our teacher")

# The Logic: You correctly handle the two scenarios:
# Scenario A: The vowel is new (not in keys) ---> Set value to 1
# Scenario B: The vowel exists (in keys) ---> Add 1 to the existing value.
```

```python
# input a message and then count a frequncy each vowel present in the message
# Same example with different code

def cv(message):
    message = message.upper()
    result = {} 
    
    for char in message:
        # First, just check if it is a vowel
        if char in "AEIOU":
            # Then check if it is already in our dictionary
            if char in result:
                result[char] += 1
            else:
                result[char] = 1
                
    print(result)

cv("Hello, Harshit Master is our teacher")
```
