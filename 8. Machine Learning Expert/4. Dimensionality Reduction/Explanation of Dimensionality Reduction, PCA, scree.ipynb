Let’s build this from **zero**, because “Dimensionality Reduction + PCA + Scree Plot” sounds like something invented by statisticians to keep students humble. The ideas are actually quite intuitive once you visualize what is happening.

# 1. What is Dimensionality Reduction?
```
Suppose your dataset has **10 features**:

```text
Age
Income
Education
Loan Amount
Credit Score
Experience
Debt
Savings
Assets
Monthly Expenses
```

So each person is represented using **10 dimensions**.

In machine learning, more features can sometimes be useful, but they can also create problems:

* More computation
* More noise
* Redundant information
* Harder visualization
* **Curse of dimensionality**
* Sometimes worse model performance

So we may want to reduce:

```text
10 dimensions → 2 or 3 dimensions
```

while trying to preserve as much **important information** as possible.

That's **dimensionality reduction**.

---

# 2. Simple real-life example

Imagine you have information about students:

| Student | Math | Physics | Chemistry |
| ------- | ---: | ------: | --------: |
| A       |   90 |      88 |        92 |
| B       |   80 |      82 |        79 |
| C       |   50 |      55 |        52 |
| D       |   95 |      91 |        94 |

We have **3 dimensions**:

```text
Math
Physics
Chemistry
```

But notice something interesting.

Students who are good at Math are generally also good at Physics and Chemistry.

So these three columns contain **similar information**.

Instead of saying:

> "Student A has Math = 90, Physics = 88, Chemistry = 92"

we could create a new variable:

```text
Overall Scientific Ability = combination of Math + Physics + Chemistry
```

Now:

```text
3 features → 1 feature
```

while retaining much of the useful information.

That's the basic intuition behind PCA.

---

# 3. Where does PCA come in?

There are many dimensionality-reduction techniques.

One of the most famous is:

> **PCA = Principal Component Analysis**

PCA takes your original features and creates **new features called Principal Components**.

For example:

```text
Original features

X1 = Math
X2 = Physics
X3 = Chemistry
```

PCA might create:

```text
PC1
PC2
PC3
```

These are not original columns.

They are **combinations of the original columns**.

For example, conceptually:

```text
PC1 = 0.58(Math) + 0.57(Physics) + 0.58(Chemistry)
```

So PC1 captures something like:

> "Overall scientific performance"

PC2 might capture a different pattern:

> "Math vs Physics difference"

PC3 captures another smaller pattern.

---

# 4. The most important idea: PCA looks for VARIANCE

This is the heart of PCA.

PCA asks:

> **"In which direction does the data vary the most?"**

Let's visualize.

Imagine your data looks like this:

```text
        •
      •
    •
  •
•
```

The points have a strong diagonal pattern.

There are many possible directions we could draw through the data.

But PCA finds the direction where the data varies the most.

```text
          •
        •
      •
    •
  •
───────────────→
       PC1
```

That direction becomes:

### PC1 = Principal Component 1

It captures the **maximum variance**.

Then PCA finds another direction perpendicular to PC1 that captures the next highest variance:

```text
          ↑ PC2
          |
          |
          |    •
          |  •
          |•
──────────┼────────→ PC1
```

So:

```text
PC1 → maximum variance
PC2 → next highest variance
PC3 → next highest variance
...
```

This is why they're called **principal components**.

---

# 5. What does "variance" mean here?

Variance basically tells us:

> **How spread out the data is.**

Imagine these two datasets:

### Dataset A

```text
48 49 50 51 52
```

Very little variation.

### Dataset B

```text
10 30 50 70 90
```

Large variation.

PCA considers directions with greater variation more informative.

Why?

Because a direction with almost no variation doesn't tell us much about how observations differ.

---

# 6. PCA is basically rotating the coordinate system

This is one of the best ways to understand PCA.

Suppose your original axes are:

```text
Y
↑
|
|
|
└────────→ X
```

Your data might be diagonally distributed:

```text
Y
↑
|        •
|      •
|    •
|  •
|•
└────────────→ X
```

PCA effectively **rotates the axes**:

```text
             PC1
            /
           /
          /  •
        •
      •
    •
  /
 / PC2
```

Now:

```text
PC1 = direction of maximum variation
PC2 = direction of remaining variation
```

This rotation makes the important structure easier to represent.

---

# 7. Why can we remove dimensions?

Suppose PCA gives us:

```text
PC1 → 80% variance
PC2 → 15% variance
PC3 → 4% variance
PC4 → 1% variance
```

Total:

```text
80 + 15 + 4 + 1 = 100%
```

Suppose we keep only:

```text
PC1
PC2
```

Then we retain:

```text
80 + 15 = 95%
```

of the variance.

So:

```text
4 original dimensions
        ↓
     PCA
        ↓
2 dimensions
        ↓
95% variance retained
```

That's dimensionality reduction.

---

# 8. What exactly is a Principal Component?

A Principal Component is a **new feature created by combining the original features**.

Suppose:

```text
X1 = Age
X2 = Income
X3 = Loan Amount
```

PCA could create:

```text
PC1 =
0.60 × Age
+ 0.65 × Income
+ 0.46 × Loan Amount
```

And:

```text
PC2 =
-0.50 × Age
+ 0.70 × Income
- 0.51 × Loan Amount
```

These numbers are called **loadings**.

So PCA doesn't simply delete columns.

It creates a **new coordinate system**.

---

# 9. A very important distinction

Suppose you have:

```text
Age
Income
LoanAmount
CreditScore
```

After PCA you might have:

```text
PC1
PC2
PC3
PC4
```

You should NOT think:

```text
PC1 = Age
PC2 = Income
PC3 = LoanAmount
```

No.

Each PC can contain information from **multiple original variables**.

Think:

```text
Original variables
       ↓
     PCA
       ↓
New combinations
       ↓
PC1, PC2, PC3...
```

---

# 10. What is Eigenvalue?

Now we're entering the mathematical machinery behind PCA.

Don't panic. Mathematics has merely arrived wearing a lab coat.

Each Principal Component has an **eigenvalue**.

The eigenvalue tells us approximately:

> **How much variance that principal component captures.**

For example:

| Component | Eigenvalue |
| --------- | ---------: |
| PC1       |        4.8 |
| PC2       |        2.1 |
| PC3       |        0.8 |
| PC4       |        0.3 |

Total:

```text
4.8 + 2.1 + 0.8 + 0.3 = 8
```

Explained variance:

```text
PC1 = 4.8 / 8 = 60%
PC2 = 2.1 / 8 = 26.25%
PC3 = 0.8 / 8 = 10%
PC4 = 0.3 / 8 = 3.75%
```

So:

```text
PC1 → 60%
PC2 → 26.25%
PC3 → 10%
PC4 → 3.75%
```

---

# 11. What is Explained Variance Ratio?

In Python, you'll often see:

```python
pca.explained_variance_ratio_
```

It tells you the **percentage of total variance explained by each principal component**.

Example:

```python
pca.explained_variance_ratio_
```

Output:

```text
[0.60, 0.2625, 0.10, 0.0375]
```

Interpretation:

```text
PC1 → 60%
PC2 → 26.25%
PC3 → 10%
PC4 → 3.75%
```

And:

```python
pca.explained_variance_ratio_.sum()
```

would give:

```text
1.0
```

because together they explain 100% of the variance.

---

# 12. Now comes the SCREE PLOT

This is probably the part you're asking about when you say **"PCA & scree."**

A **scree plot** is a graph showing how much variance each principal component explains.

For example:

```text
Variance
  |
60| ●
  |  \
50|   \
  |    \
40|     \
  |      ●
30|       \
  |        \
20|         ●
  |          \
10|           ●
  |            \____
  |
  +--------------------
     PC1 PC2 PC3 PC4
```

Typically:

```text
X-axis → Principal Components
Y-axis → Explained Variance
```

---

# 13. Why is it called a Scree Plot?

"Scree" basically refers to **loose rocks at the bottom of a mountain**.

The graph often looks like:

```text
      ●
       \
        ●
         \
          ●
           \
            ●__________
```

There's usually a steep decline followed by a flatter region.

The point where the curve starts flattening is called the:

> **Elbow**

Humans apparently needed a geological metaphor for deciding how many columns to keep.

---

# 14. How do we use the Scree Plot?

Suppose your PCA results are:

```text
PC1 → 50%
PC2 → 25%
PC3 → 15%
PC4 → 5%
PC5 → 3%
PC6 → 2%
```

Graph:

```text
Variance
  |
50| ●
  |  \
40|   \
  |    \
30|     \
  |      ●
20|       \
  |        ●
10|         \
  |          ●
  |           ●──●
  +------------------
    1  2  3  4  5  6
```

The curve becomes relatively flat after PC3.

So you might choose:

```text
PC1
PC2
PC3
```

because they capture:

```text
50 + 25 + 15 = 90%
```

of the variance.

---

# 15. Scree Plot vs Cumulative Explained Variance

These are related but different.

### Scree plot

Shows:

```text
Individual contribution

PC1 → 50%
PC2 → 25%
PC3 → 15%
PC4 → 5%
...
```

### Cumulative explained variance

Shows:

```text
PC1 → 50%
PC1 + PC2 → 75%
PC1 + PC2 + PC3 → 90%
PC1 + ... + PC4 → 95%
```

You can plot cumulative variance:

```text
100% |                    ●
 90% |              ●
 80% |
 70% |        ●
 60% |
 50% |   ●
     +-----------------------
       1   2   3   4
          Components
```

If your target is **95% variance**, you choose enough PCs to reach approximately 95%.

# 16. How PCA works step-by-step

At a high level:

```text
              Original Dataset
                     ↓
          Select numerical features
                     ↓
             Handle missing values
                     ↓
                 Scale data
                     ↓
              Calculate covariance
                     ↓
       Find eigenvectors/eigenvalues
                     ↓
         Create Principal Components
                     ↓
       Calculate explained variance
                     ↓
       Decide how many PCs to keep
                     ↓
          Transform the dataset
```

In practice, libraries like `scikit-learn` do the mathematics for you.

# 17. Why do we need StandardScaler before PCA?

This is **very important**.

Suppose your dataset has:

```text
Age          → 18–80
Income       → 10,000–500,000
LoanAmount   → 50,000–10,000,000
```

The scales are completely different.

PCA is variance-based.

Therefore, a feature with a huge numerical scale can dominate the calculation.

So normally:

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

Then:

```python
from sklearn.decomposition import PCA

pca = PCA()

X_pca = pca.fit_transform(X_scaled)
```

Typical pipeline:

```text
X
↓
StandardScaler
↓
X_scaled
↓
PCA
↓
X_pca
```

# 18. A tiny Python example

Suppose:

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

pca = PCA()

X_pca = pca.fit_transform(X_scaled)
```

Now:

```python
pca.explained_variance_ratio_
```

might produce:

```text
[0.52, 0.27, 0.12, 0.06, 0.03]
```

Meaning:

```text
PC1 → 52%
PC2 → 27%
PC3 → 12%
PC4 → 6%
PC5 → 3%
```

Cumulative:

```python
import numpy as np

np.cumsum(pca.explained_variance_ratio_)
```

Output:

```text
[0.52, 0.79, 0.91, 0.97, 1.00]
```

So:

```text
1 PC  → 52%
2 PCs → 79%
3 PCs → 91%
4 PCs → 97%
5 PCs → 100%
```

If you want approximately **95%**, you need **4 components**.

---

# 19. Even easier: let PCA choose the number

You can directly tell PCA:

```python
pca = PCA(n_components=0.95)
```

This means:

> "Keep enough principal components to preserve at least 95% of the variance."

Then:

```python
X_pca = pca.fit_transform(X_scaled)
```

If the original dataset had:

```text
20 features
```

and PCA determines that:

```text
8 components
```

are enough for 95% variance, then:

```text
20 features
      ↓
     PCA
      ↓
8 components
```

That's a substantial dimensionality reduction.

# 21. One major limitation of PCA

PCA creates new variables:

```text
PC1
PC2
PC3
```

These are combinations of original features.

Therefore, interpretability can become difficult.

For example:

```text
PC1 =
0.42 Age
+ 0.51 Income
- 0.31 Debt
+ 0.63 LoanAmount
...
```

What does PC1 actually *mean*?

Not always obvious.

So PCA gives you:

**Less dimensions + potentially less noise + computational benefits**

but sometimes at the cost of:

**Interpretability.**

# 23. The 5 things you should remember for your exam/interview

### 1. Dimensionality Reduction

> Reducing the number of features while retaining as much useful information as possible.

### 2. PCA

> PCA is a dimensionality-reduction technique that transforms original features into new, uncorrelated principal components.

### 3. PC1

> PC1 captures the **maximum variance** in the data.

### 4. PC2

> PC2 captures the **next highest variance**, subject to being orthogonal to PC1.

And so on.

### 5. Scree Plot

> A scree plot shows the variance explained by each principal component and helps determine how many components to retain.
