# Rules for creating Variables

* Variable names must start with a letter (a-z or A-2) or an underscore (.). They cannot begin with a number.)

* Variable names can consist of letters, numbers, and underscores. They are case-sensitive, meaning that "myVar" and "myvar" are considered different variables.

* It is good practice to use descriptive names that reflect the purpose or meaning of the variable. For example, "name," "age," or "total score" are more meaningful than generic names like "x" or "y."
  
* Avoid using reserved keywords as variable names. Reserved keywords are words that have special meanings in Python and cannot be used as variable names. Examples of reserved keywords include "if," "for," "while," and "print."

```python
# Code Blocks and Nested Block
def greet(name):
    # Outer code block
    print("Hello, "+ name + "!")
    if name == "Alice":
        # Nested code block 
        print("You are amazing, " + name + "!")
    else:
        # Another nested code block
        print("Nice to meet you, " + name + "!")
    # Code outside the if-else block 
    print("Have a great day, " + name + "!")
# Function call
greet("Rohit")
```

# Sets

A Set in Python is a collection of unique elements that is unordered and mutable.

Think of it like a bag of distinct items where the order you put them in doesn't matter, and duplicates are automatically thrown out.

## Key Features
* Unique Elements: Sets do not allow duplicate values. If you try to add a duplicate, it is ignored.

* Unordered: Items have no fixed position. You cannot access items using an index (e.g., myset[0] will raise an error).

* Mutable: You can add or remove items after the set is created.

* Syntax: Defined using curly braces {} or the set() function.

```python
# Curly brackets and built in set() function
```

```python
# Creating a set using curly breackets
fruits = {"apple", "banana", "cherry"}
print(fruits)
# Creating a set using a set() function
colors = set(['red', 'blue', 'green'])
print(colors)
```

```python
# Set Operations
# Creating two sets 
set1 = {1, 2, 3, 4, 5}
set2 = { 4, 5, 6, 7, 8}
# Union of two sets
union_set = set1.union(set2)
print(union_set)
# Intersection of two sets
intersection_set = set1.intersection(set2)
print(intersection_set)
# Difference between two sets
difference_set = set1.difference(set2)
print(difference_set)
# Checking if a set is a subset of another set
subset_check = set1.issubset(set2)
print(subset_check)
```

```python
# Modifying sets
# Creating a set
numbers = {1, 2, 3}
# Adding elements to a set
numbers.add(4)
print(numbers)
# Removing an element from a set
numbers.remove(2)
print(numbers)
```

```python
# Frozensets : is an immutable version of set

# Creating a frozenset
fs = frozenset([1, 2, 3])
print(fs)
```

```python
# Set Comprehension - The set comprehension iterates over a range of numbers and calcualtes the square of each number, resulting in a set of squares

squares = {x**2 for x in range(1, 6)}
print(squares)
```

# Lambdas : Anonymous Functions

Lambdas, provide a concise way to define small, one-line functions without a formal function name, Useful in situations where a function is required for a short and specific purpose.

Syntax : 
lambda arguments: expression

lambda keyword is followed by a list of arguments which represets the inputs to the lambda.

```python
# Example of lambda function that calculates the square of a number:

square = lambda x: x ** 2
```

# Using Lambdas:

```python
# Sorting : Lambdas can be used as key functions in sorting operations to define custom sorting criteria

fruits = ['apple', 'banana', 'cherry', 'date']
sorted_fruits = sorted(fruits, key = lambda x: len(x)) # the lambda function 'lambda x: len(x)' is used as the key function to sort the list of fruits based on their lengths 
print(sorted_fruits)

# Filtering : Lambdas can be used in combination with built-in functions like 'filter()' to filter elements from a list based on specific conditions
#Example : filters even numbers from a list 

numbers = [1, 2, 3, 4, 5, 6]
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)

# Mapping : Lambdas can be used with functions like 'map()' to apply a specific operation to each element of a list,
#Example : Doubles each number in a list 

numbers = [1, 2, 3, 4, 5]
doubled_numbers = list(map(lambda x: x * 2, numbers))
print(doubled_numbers)
```

```python
# Example : Consider a scenario where you have a list of students, and you need to sort them based on their names.

students = [
    {'name': 'Alice', 'age': 20},
    {'name': 'Bob', 'age': 19},
    {'name': 'Charlie', 'age': 21}
]
sorted_students = sorted(students, key=lambda x: x['name'])
print(sorted_students)

# in this example, lambda function 'lambda x: x['name'] is used as the key function to sort the list of students based on their names
```

# Understanding Scope of Variables in Functions

The concept of variable scope defines the accessibility and lifespan of variables within a program

1. Local Scope : When a vairable is defined inside a function, it has local scope. This means that the variable is only accessible within the funtion where it is defiend.

```python
# Ex

def my_function():
    x = 10
    print(x)
my_function()
# print(x) # Error : NameError: name 'x' is not defined

# Local variables are created when the function is called and destroyed when the function completes execution. They are not accessible outside the funtion
```

2. Global Scope : Variables defined outside any function or block have global scope, meaning they can be accessed from anywhere in the program.

```python
# Ex

x = 10
def my_function():
    print(x)
my_function()
print(x)

# Global vairables can be accessed and modified from any part of the program
```

```python
# However, modifying global variables within a function requires explicitly decalring them as global using the 'global' keyword.
```

```python
# Modifying Global Variables within a function
#Ex

x = 10
def modify_global():
    global x
    x = 20
modify_global()
print(x)

# It is generally considered good practice to minimize the use of global variables.
```

```python
'''Example : Consider a scenario where you have a global variable 'total' representing the total sales of a company. You want to calculate the bonus
amount for each  employee based on their individual sales, while keeping track of the overall sales total.
'''

total = 0
def calculate_bonus(sales):
    global total
    bonus = sales * 0.1 # 10 % of bonus
    total += sales # total = total + sales
    return bonus
employee1_sales = 5000
employee1_bonus = calculate_bonus(employee1_sales)
print("Employee 1 Bonus:", employee1_bonus)
print("Total Sales:", total)
```
