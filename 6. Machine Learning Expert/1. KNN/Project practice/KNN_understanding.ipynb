K = how many neighbors we consider
Nearest = closest according to some distance
Neighbors = existing data points
3. But how do we know which point is closest?

We calculate distance.

For two-dimensional data, we use Euclidean distance.

The formula is:

$$ d = \sqrt{(x_2-x_1)^2+(y_2-y_1)^2} $$

Here:

x = Age
y = Glucose
4. Calculate distance manually

Our new patient:

Age = 32
Glucose = 115

Let's calculate the distance to patient C.

Patient C:

Age = 35
Glucose = 120

Therefore:

$$ d = \sqrt{(35-32)^2+(120-115)^2} $$ $$ d = \sqrt{3^2+5^2} $$ $$ d = \sqrt{9+25} $$ $$ d = \sqrt{34} $$ $$ d \approx 5.83 $$

So:

Distance from New Patient → C = 5.83

Now calculate distance to B

Patient B:

Age = 28
Glucose = 105
$$ d = \sqrt{(28-32)^2+(105-115)^2} $$ $$ d = \sqrt{(-4)^2+(-10)^2} $$ $$ d = \sqrt{16+100} $$ $$ d = \sqrt{116} $$ $$ d \approx 10.77 $$

So:

Distance → B = 10.77

Therefore:

C = 5.83   ← closer
B = 10.77
5. KNN does this for EVERY training point

This is an important part.

KNN doesn't magically know which points are nearby.

For a new observation, it calculates the distance to all training observations.

8. What if K = 1?

Now we only look at the single closest patient.

Our closest is:

C → No

Therefore:

Prediction = No

This is called:

K = 1

The model trusts one single neighbor.

That can be dangerous because if that one neighbor happens to be unusual or noisy, our prediction can be wrong.

9. Why not always use a huge K?

Suppose:

k = 100

Now we're considering a huge number of observations.

The prediction becomes influenced by points that may be very far away.

The model becomes less sensitive to the local neighborhood.

So we have a trade-off.

Small K
K = 1
K = 2
K = 3

More sensitive to nearby observations.

Can lead to:

Overfitting

Large K
K = 50
K = 100
K = 200

More generalized.

Can lead to:

Underfitting

Conceptually:

Small K
   ↓
Very local decisions
   ↓
Can overfit


Large K
   ↓
Very broad decisions
   ↓
Can underfit

That's why we'll eventually find a suitable K instead of arbitrarily declaring that 5 is the sacred number.
